```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv(
    'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv'
)

tips
```

### Step 4. Delete the Unnamed 0 column

```python
tips = tips.drop('Unnamed: 0', axis=1)
```

### Step 5. Plot the total_bill column histogram

```python
plt.hist(tips['total_bill'])

plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.title('Total Bill Histogram')

plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.scatter(tips['total_bill'], tips['tip'])

plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')

plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.pairplot(tips[['total_bill', 'tip', 'size']])

plt.show()
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.boxplot(x='day', y='total_bill', data=tips)

plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.title('Total Bill by Day')

plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.scatterplot(
    x='tip',
    y='day',
    hue='sex',
    data=tips
)

plt.xlabel('Tip')
plt.ylabel('Day')
plt.title('Tip by Day and Sex')

plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.scatterplot(
    x='tip',
    y='day',
    hue='sex',
    data=tips
)

plt.xlabel('Tip')
plt.ylabel('Day')
plt.title('Tip by Day and Sex')

plt.show()
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
sns.boxplot(
    x='day',
    y='total_bill',
    hue='time',
    data=tips
)

plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.title('Total Bill by Day and Time')

plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.scatterplot(
    data=tips[tips['sex'] == 'Male'],
    x='total_bill',
    y='tip',
    hue='smoker',
    ax=axes[0]
)

axes[0].set_title('Male')
axes[0].set_xlabel('Total Bill')
axes[0].set_ylabel('Tip')


sns.scatterplot(
    data=tips[tips['sex'] == 'Female'],
    x='total_bill',
    y='tip',
    hue='smoker',
    ax=axes[1]
)

axes[1].set_title('Female')
axes[1].set_xlabel('Total Bill')
axes[1].set_ylabel('Tip')

plt.show()
```
