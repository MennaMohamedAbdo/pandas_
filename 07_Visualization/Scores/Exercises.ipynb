### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2. Create the DataFrame that should look like the one below.

```python
data = {
    'firstName': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy', 'Neil'],
    'lastName': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze', 'Eason'],
    'preTestScore': [4.0, 24.0, 31.0, 2.0, 3.0, 22.0],
    'postTestScore': [25.0, 94.0, 57.0, 62.0, 70.0, 88.0],
    'age': [19, 18, 19, 20, 20, 19],
    'sex': ['F', 'F', 'F', 'M', 'M', 'M']
}

df = pd.DataFrame(data)

df
```

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

```python
plt.scatter(
    df['preTestScore'],
    df['postTestScore'],
    s=df['age'] * 10
)

plt.xlabel('Pre Test Score')
plt.ylabel('Post Test Score')
plt.title('Pre Test Score vs Post Test Score')

plt.show()
```

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

```python
colors = df['sex'].map({'F': 'pink', 'M': 'blue'})

plt.scatter(
    df['preTestScore'],
    df['postTestScore'],
    s=df['postTestScore'] * 4.5,
    c=colors
)

plt.xlabel('Pre Test Score')
plt.ylabel('Post Test Score')
plt.title('Pre Test Score vs Post Test Score by Sex')

plt.show()
```
