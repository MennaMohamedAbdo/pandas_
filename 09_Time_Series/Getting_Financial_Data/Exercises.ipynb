```python
import requests
import pandas as pd

url = "https://www.alphavantage.co/query"

params = {
    "function": "TIME_SERIES_DAILY",
    "symbol": "AAPL",
    "outputsize": "compact",
    "apikey": api_key
}

response = requests.get(url, params=params)
data = response.json()

df_apple = pd.DataFrame.from_dict(
    data["Time Series (Daily)"],
    orient="index"
)

df_apple.index = pd.to_datetime(df_apple.index)
df_apple = df_apple.sort_index()

df_apple.columns = ['open', 'high', 'low', 'close', 'volume']

print(df_apple)
```
